```python
# GPU 테스트
import numpy as np
import lightgbm as lgb

X = np.array([[1,2],[3,4]])
y = np.array([1,0])

params = {
    "objective": "regression",
    "device": "gpu"
}

train_data = lgb.Dataset(X, label=y)
bst = lgb.train(params, train_data, num_boost_round=2)

print("GPU test OK")

```

```python
import lightgbm as lgb
import numpy as np

# 데이터 준비 (예시)
X = np.random.rand(1000, 10)
y = np.random.rand(1000)

# GPU 활성화를 위한 파라미터 설정
params = {
    'objective': 'regression',
    'metric': 'rmse',
    'device': 'gpu',  # <-- 이 파라미터가 GPU 사용을 지시합니다.
    'random_state': 42
}

# 모델 학습
gbm = lgb.LGBMRegressor(**params)
gbm.fit(X, y)

print("LightGBM 모델 학습 완료. GPU를 사용했는지 확인하세요.")
```
