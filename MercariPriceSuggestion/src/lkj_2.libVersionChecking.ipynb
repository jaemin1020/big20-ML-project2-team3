```python
# GPU 테스트
import numpy as np
import lightgbm as lgb

X = np.array([[1,2],[3,4]])
y = np.array([1,0])

params = {
    "objective": "regression",
    "device": "gpu"
}

train_data = lgb.Dataset(X, label=y)
bst = lgb.train(params, train_data, num_boost_round=2)

print("GPU test OK")

```
