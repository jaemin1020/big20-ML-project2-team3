```python
import torch

# 현재 GPU 개수 확인
print("GPU 개수:", torch.cuda.device_count())

# 현재 선택된 GPU 인덱스
print("현재 GPU 인덱스:", torch.cuda.current_device())

# GPU 메모리 사용량 확인
print("GPU 메모리 사용량:",
      torch.cuda.memory_allocated(0) / 1024**2, "MB")
```
