```python
import os

def find_utils_folder(start_path):
    """utils 폴더를 찾는 함수"""
    for root, dirs, files in os.walk(start_path):
        if 'utils' in dirs:
            utils_path = os.path.join(root, 'utils')
            has_init = os.path.exists(os.path.join(utils_path, '__init__.py'))
            print(f"✓ Found utils at: {utils_path}")
            print(f"  Has __init__.py: {has_init}")
            if has_init:
                print(f"  Files inside:")
                for f in os.listdir(utils_path):
                    print(f"    - {f}")
            return utils_path
    return None

# 현재 위치부터 2단계 위까지 검색
current = os.getcwd()
parent = os.path.dirname(current)
grandparent = os.path.dirname(parent)

print("Searching for utils folder...\n")
utils_found = find_utils_folder(grandparent)

if utils_found:
    project_root = os.path.dirname(utils_found)
    print(f"\n✓ Project root should be: {project_root}")
    
    import sys
    if project_root not in sys.path:
        sys.path.insert(0, project_root)
    
    print("\nTrying to import...")
    try:
        from utils import preprocessing
        print("✓ Import successful!")
    except Exception as e:
        print(f"✗ Import failed: {e}")
else:
    print("\n✗ utils folder not found!")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt

# 예시 데이터 (산탄데르 고객만족도 데이터 일부 가정)
data = pd.DataFrame({
    'satisfaction': [3, 4, 2, 5, 1],
    'service_quality': [4, 5, 3, 5, 2],
    'response_time': [2, 3, 1, 4, 1],
    'staff_friendliness': [5, 4, 3, 5, 2]
})

# 1. 데이터 표준화
scaler = StandardScaler()
data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)

# 2. 상관행렬 계산
correlation_matrix = data_scaled.corr()

# 3. 히트맵 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('산탄데르 고객만족도 상관행렬 히트맵')
plt.tight_layout()
plt.show()
```
