```python
myCols = [
    "ind_var37_cte", "ind_var37_0", "ind_var37", "ind_var39_0", "ind_var40_0",
    "ind_var40", "ind_var41_0", "ind_var41", "ind_var39", "ind_var44_0",
    "ind_var44", "ind_var46_0", "ind_var46", "num_var1_0", "num_var1",
    "num_var4", "num_var5_0", "num_var5", "num_var6_0", "num_var6",
    "num_var8_0", "num_var8", "num_var12_0", "num_var12", "num_var13_0",
    "num_var13_corto_0", "num_var13_corto", "num_var13_largo_0",
    "num_var13_largo", "num_var13_medio_0", "num_var13_medio", "num_var13",
    "num_var14_0", "num_var14", "num_var17_0", "num_var17", "num_var18_0",
    "num_var18", "num_var20_0", "num_var20", "num_var24_0", "num_var24",
    "num_var26_0", "num_var26", "num_var25_0", "num_var25",
    "num_op_var40_hace2", "num_op_var40_hace3", "num_op_var40_ult1",
    "num_op_var40_ult3", "num_op_var41_hace2", "num_op_var41_hace3",
    "num_op_var41_ult1", "num_op_var41_ult3", "num_op_var39_hace2",
    "num_op_var39_hace3", "num_op_var39_ult1", "num_op_var39_ult3",
    "num_var27_0", "num_var28_0", "num_var28", "num_var27", "num_var29_0",
    "num_var29", "num_var30_0", "num_var30", "num_var31_0", "num_var31",
    "num_var32_0", "num_var32", "num_var33_0", "num_var33", "num_var34_0",
    "num_var34"
]

allCols = df.columns[1:-1].to_list()
allCols
```

```python
# 방법 2: 요약 정보 (DataFrame으로)
print(f'\n{"="*102}')
print("전체 요약".center(102))
print("="*102)

summary_data = []
for col in myCols:
    if col in df.columns:
        zero_count = (df[col] == 0).sum()
        mode_freq = df[col].value_counts().iloc[0] if len(df[col]) > 0 else 0

        summary_data.append({
            'ColumnName': col,  # 컬럼명
            'na_Sum': df[col].isna().sum(),  # 결측치_개수
            'nUnique': df[col].nunique(),  # 고유값_개수
            'mode': df[col].mode()[0] if len(df[col].mode()) > 0 else None,  # 최빈값
            'modeFreq': mode_freq,  # 최빈값_빈도
            'modeFreqRate': f"{(mode_freq / len(df) * 100):.2f}%",  # 최빈값_빈도_비율
            'zero_count': zero_count,  # 0값_개수
            'zero_count_rate': f"{(zero_count / len(df) * 100):.2f}%"  # 0값_비율
        })

summary_df = pd.DataFrame(summary_data)
# summary_df.to_csv('../doc/data_summary.csv')
# allCols

summary_sorted = summary_df.sort_values('zero_count', ascending=False)
print(summary_sorted.to_string(index=False))
```

```python
# -----------------------------------------
# 1. 기본: 99.5% 이상 zero인 컬럼
# -----------------------------------------
high_zero_cols = summary_df[
    summary_df['zero_count'] / len(df) >= 0.995
]['ColumnName'].tolist()


# -----------------------------------------
# 2. 고유값이 1개인(무조건 useless) 컬럼
# -----------------------------------------
constant_cols = summary_df[
    summary_df['nUnique'] == 1
]['ColumnName'].tolist()


# -----------------------------------------
# 3. 완전히 중복된 컬럼 자동 탐지
# -----------------------------------------
duplicate_cols = []
seen = {}

for col in myCols:
    if col in df.columns:
        series = tuple(df[col].values)   # 값 전체를 튜플로 변환
        if series in seen:
            duplicate_cols.append(col)
        else:
            seen[series] = col


# -----------------------------------------
# 4. 최종 삭제 후보 묶기
# -----------------------------------------
drop_candidates = sorted(set(high_zero_cols + constant_cols + duplicate_cols))

print(f"삭제해도 안전한 컬럼 개수: {len(drop_candidates)}")
print(drop_candidates)
```

```python
# ===============================================
# 1. 삭제해도 안전한 컬럼 목록
# ===============================================
drop_candidates = [
    'ind_var37', 'ind_var39', 'ind_var40', 'ind_var41', 'ind_var44', 'ind_var44_0',
    'ind_var46', 'ind_var46_0', 'num_op_var40_hace2', 'num_op_var40_hace3',
    'num_op_var40_ult1', 'num_op_var40_ult3', 'num_var1', 'num_var13_medio',
    'num_var13_medio_0', 'num_var17', 'num_var17_0', 'num_var18', 'num_var18_0',
    'num_var20', 'num_var20_0', 'num_var25', 'num_var26', 'num_var27', 'num_var27_0',
    'num_var28', 'num_var28_0', 'num_var29', 'num_var29_0', 'num_var31', 'num_var31_0',
    'num_var32', 'num_var32_0', 'num_var33', 'num_var33_0', 'num_var34', 'num_var34_0',
    'num_var6', 'num_var6_0'
]

# ===============================================
# 2. myCols 내부에서만 삭제할 컬럼을 선택
# ===============================================
drop_in_myCols = sorted(set(myCols).intersection(drop_candidates))

print(f"myCols 내부에서 삭제할 컬럼 개수: {len(drop_in_myCols)}")
print(drop_in_myCols)

# ===============================================
# 3. df에서 해당 컬럼만 삭제
# ===============================================
df_clean2 = df.drop(columns=drop_in_myCols, errors='ignore')

print(f"원본 df shape : {df.shape}")
print(f"myCols 기반 삭제 후 df_clean2 shape : {df_clean2.shape}")


# ===============================================
# 4. SUMMARY 자동 생성 함수 (그대로 사용)
# ===============================================
summary_list = []

def make_summary(dataframe, col_list):
    summary_list = []

    # dataframe에 실제 존재하는 컬럼만 사용
    valid_cols = [c for c in col_list if c in dataframe.columns]

    for col in valid_cols:
        zero_count = (dataframe[col] == 0).sum()
        mode_val = dataframe[col].mode()[0] if len(dataframe[col].mode()) > 0 else None
        mode_freq = dataframe[col].value_counts().iloc[0]

        summary_list.append({
            'ColumnName': col,
            'na_Sum': dataframe[col].isna().sum(),
            'nUnique': dataframe[col].nunique(),
            'mode': mode_val,
            'modeFreq': mode_freq,
            'modeFreqRate': f"{mode_freq / len(dataframe) * 100:.2f}%",
            'zero_count': zero_count,
            'zero_count_rate': f"{zero_count / len(dataframe) * 100:.2f}%"
        })

    return pd.DataFrame(summary_list)

# ===============================================
# 5. 정제 후 SUMMARY 생성
# ===============================================
summary_clean_df = make_summary(df_clean2, myCols)
summary_clean_sorted = summary_clean_df.sort_values('zero_count', ascending=False)
print("\n==================== 정제 후 요약 ====================")
print(summary_clean_sorted.to_string(index=False))
```
